```python
from pyspark.sql.functions import *

df_datacenter_bronze = spark.read.table("portfolio_energia.bronze.datacenter_bronze_raw")

df_datacenter_bronze.groupBy("name").count().filter(col("count") > 1).show(truncate=False)
#verifica duplicatas da granularidade name

df_nulo_datacenter = df_datacenter_bronze.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df_datacenter_bronze.columns
])
display(df_nulo_datacenter)
#verifica valores nulos nas tabelas

df_datacenter_bronze.filter(col("region").isNull()).select("name", "operator", "region").show(truncate=False)
#mostra quais datacenters vieram sem region
```
