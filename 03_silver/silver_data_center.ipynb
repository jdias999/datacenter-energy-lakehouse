```python
df_datacenter_bronze = spark.read.table("portfolio_energia.bronze.datacenter_bronze_raw")
```

```python
from pyspark.sql.functions import *


df_datacenter_silver = (
    df_datacenter_bronze
    .withColumn("id", trim(col("id")))
    .withColumn("name", trim(col("name")))
    .withColumn("operator", trim(col("operator")))
    .withColumn("region", trim(col("region")))
    .withColumn("iso2", upper(trim(col("iso2"))))
    .withColumn("ingestion_timestamp", current_timestamp())
    .dropna(subset=['id', "iso2"])
    .dropDuplicates(["id"])
    .fillna({"region": "Unknown", "operator": "Unknown", "name": "Unknown"})
)
```

```python
df_datacenter_silver.write \
    .format("delta") \
    .mode("overwrite").saveAsTable("portfolio_energia.silver.datacenter_silver")
```
